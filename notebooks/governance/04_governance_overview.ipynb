# 4. One governance view, multiple control planes

**Question:** Can one view explain the estate, its evidence, and its blind spots?

This notebook is a small reporting surface, not a new enforcement system. It combines normalized asset/evidence records and preserves source, cloud, collection time, and evidence kind. Native portals remain the authority for investigation and action.

The default view is **synthetic**. It is not a complete enterprise inventory, a compliance certification, or an Agent 365 replacement. Missing security connectors are visible; there is no misleading green 'zero risks' score.

```python
from pathlib import Path
import sys

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'src/governance').is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from IPython.display import Markdown, display
from src.governance.evidence import synthetic_estate, load_snapshot, overview, coverage, markdown_table, export_estate

# Optional normalized export from the CLI; original timestamps are preserved.
SNAPSHOT_PATH = None
estate = load_snapshot(Path(SNAPSHOT_PATH)) if SNAPSHOT_PATH else synthetic_estate()
display(Markdown(markdown_table(overview(estate))))
print('Observation coverage:', coverage(estate))
```

```python
gaps = [row.model_dump(mode='json') for row in estate.evidence if row.status in ('unknown', 'error')]
display(Markdown('## Visibility gaps and collection failures'))
display(Markdown(markdown_table(gaps)))
print('A snapshot remains a snapshot, even if displayed during a live presentation.')
```

## Portable reporting boundary

`export_estate` writes normalized JSON and a flat overview CSV. A fresh output directory prevents accidental overwrite. Collectors export selected metadata, not raw responses, keys, connection strings, prompts, or completions. Resource names and IDs are still organizational metadata; exports are local and git-ignored by default.

The CSV is a possible reporting input, not a deployed Power BI dashboard. Cross-cloud transport, permissions, data classification, licensing, and refresh are outside this demo. Native source links provide drill-through; the overview does not control those systems.

```python
WRITE_LOCAL_EXPORT = False

if WRITE_LOCAL_EXPORT:
    from datetime import datetime, timezone
    run_id = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
    output = ROOT / 'notebooks/governance/exports' / run_id
    for path in export_estate(estate, output):
        print(path.relative_to(ROOT))
else:
    print('Export disabled. Notebook execution leaves no evidence files behind.')
```
